```python
import nltk
import re

from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()
```

```python
feedback = """
I am very happy with the product. The quality is excellent
and the delivery was fast. The customer service was also very
helpful. I would definitely recommend this product to others.
"""
```

```python
# Sentiment analysis
score = sia.polarity_scores(feedback)

if score['compound'] >= 0.05:
    sentiment = "Positive 😊"
elif score['compound'] <= -0.05:
    sentiment = "Negative 😞"
else:
    sentiment = "Neutral 😐"

# Word count
words = re.findall(r'\b\w+\b', feedback)
word_count = len(words)

# Important keywords
keywords = [
    "product",
    "quality",
    "price",
    "delivery",
    "service",
    "customer",
    "support",
    "staff",
    "experience",
    "recommend"
]

found_keywords = []

for keyword in keywords:
    if keyword.lower() in feedback.lower():
        found_keywords.append(keyword)

# Display result
print("=" * 55)
print("          CUSTOMER FEEDBACK ANALYSIS SYSTEM")
print("=" * 55)

print("\nCUSTOMER FEEDBACK:")
print(feedback)

print("\n--- SENTIMENT ANALYSIS ---")
print("Positive Score :", round(score['pos'], 2))
print("Negative Score :", round(score['neg'], 2))
print("Neutral Score  :", round(score['neu'], 2))
print("Compound Score :", round(score['compound'], 2))
print("Overall Sentiment:", sentiment)

print("\n--- FEEDBACK STATISTICS ---")
print("Word Count:", word_count)

print("\n--- IMPORTANT KEYWORDS ---")
print(found_keywords if found_keywords else "No keywords found")

print("\n--- CUSTOMER SATISFACTION ---")

if sentiment.startswith("Positive"):
    print("👍 Customer is satisfied with the product/service.")
    print("Recommendation: Continue the current service quality.")

elif sentiment.startswith("Negative"):
    print("⚠️ Customer is not satisfied.")
    print("Recommendation: Improve product/service quality.")

else:
    print("ℹ️ Customer has a neutral opinion.")
    print("Recommendation: Collect more feedback.")

print("\n" + "=" * 55)
```
